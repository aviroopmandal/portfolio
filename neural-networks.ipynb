## Cell 2: Import libraries and create the dataset

```python
import matplotlib.pyplot as plt
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim

X = torch.linspace(-3, 3, 1000).unsqueeze(1)
y = X**2 + torch.randn_like(X) * 0.5

X_train, X_test = X[:800], X[800:]
y_train, y_test = y[:800], y[800:]

plt.scatter(X_train.numpy(), y_train.numpy(), alpha=0.6, s=10)
plt.title("Our Data: A Noisy U-Shape Curve")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## Cell 3: Define and instantiate the neural network

```python
class TinyBrain(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(1, 20)
        self.layer2 = nn.Linear(20, 20)
        self.layer3 = nn.Linear(20, 1)
        self.activation = nn.ReLU()

    def forward(self, x):
        x = self.activation(self.layer1(x))
        x = self.activation(self.layer2(x))
        return self.layer3(x)


model = TinyBrain()
print("Model Architecture:")
print(model)
```

## Cell 4: Configure the loss function and optimizer

```python
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
loss_history = []
```

## Cell 5: Train the model

```python
for epoch in range(500):
    predictions = model(X_train)
    loss = criterion(predictions, y_train)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 50 == 0:
        loss_history.append(loss.item())
        print(f"Epoch {epoch:3d} | Loss: {loss.item():.4f}")

print("\nTraining complete!")
```

## Cell 6: Evaluate the model and visualize its predictions

```python
model.eval()

with torch.no_grad():
    y_test_pred = model(X_test)

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.scatter(X_train.numpy(), y_train.numpy(), alpha=0.3, s=10, label="Real Data")
plt.scatter(
    X_train.numpy(),
    model(X_train).detach().numpy(),
    s=3,
    c="red",
    label="Model Prediction",
)
plt.title("Training Data: Model Fit (Red)")
plt.legend()

plt.subplot(1, 2, 2)
plt.scatter(X_test.numpy(), y_test.numpy(), alpha=0.5, s=10, label="Real Test Data")
plt.scatter(
    X_test.numpy(),
    y_test_pred.numpy(),
    s=3,
    c="red",
    label="Model Prediction",
)
plt.title("Test Data: Does it generalize?")
plt.legend()

plt.tight_layout()
plt.show()

plt.figure()
plt.plot(range(0, 500, 50), loss_history, marker="o")
plt.title("Learning Progress: Loss Decreases Over Time")
plt.xlabel("Epochs")
plt.ylabel("Loss (Error)")
plt.grid(True)
plt.show()
```
