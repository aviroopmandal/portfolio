## Cell 2

```python
import os
import getpass

os.environ["GOOGLE_API_KEY"] = getpass.getpass(
    "Enter your Gemini API key: "
)
```

## Cell 3

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0.3
)
```

## Cell 4

```python
from langchain_core.prompts import (
    ChatPromptTemplate,
    FewShotChatMessagePromptTemplate
)

from langchain_core.output_parsers import StrOutputParser
```

## Cell 5

```python
examples = [
    {
        "review": "The food was delicious!",
        "sentiment": "Positive"
    },
    {
        "review": "The service was terrible.",
        "sentiment": "Negative"
    },
    {
        "review": "The restaurant opens at 9 AM.",
        "sentiment": "Neutral"
    }
]
```

## Cell 6

```python
example_prompt = ChatPromptTemplate.from_messages([
    ("human", "{review}"),
    ("ai", "{sentiment}")
])
```

## Cell 7

```python
few_shot_prompt = FewShotChatMessagePromptTemplate(
    example_prompt=example_prompt,
    examples=examples
)
```

## Cell 8

```python
final_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
        You are a sentiment classification assistant.

        Classify the supplied text as Positive,
        Negative or Neutral.

        Also provide a one-sentence explanation
        for your classification.

        Return:
        Sentiment:
        Explanation:
        """
    ),
    few_shot_prompt,
    ("human", "{review}")
])
```

## Cell 9

```python
chain = (
    final_prompt
    | model
    | StrOutputParser()
)
```

## Cell 10

```python
review = "The product was excellent and delivery was fast!"

response = chain.invoke({
    "review": review
})

print(response)
```
