## Cell 1

```python
import torch
import torch.nn as nn

print("PyTorch version:", torch.__version__)
print("Device: CPU")
print("Feels-Like Temperature Predictor ready!")
```

## Cell 2

```python
data = [
    [25, 40, 25.5],
    [25, 60, 26.0],
    [25, 80, 26.5],
    [30, 40, 31.0],
    [30, 60, 32.0],
    [30, 80, 35.0],
    [32, 40, 33.5],
    [32, 60, 36.0],
    [32, 80, 40.0],
    [35, 40, 37.5],
    [35, 60, 41.0],
    [35, 80, 46.0]
]

print("Dataset created successfully!")
print("Number of observations:", len(data))
```

## Cell 3

```python
import numpy as np

data = np.array(data, dtype=np.float32)

X = torch.tensor(data[:, :2])
y = torch.tensor(data[:, 2:])

print("Input tensor shape:", X.shape)
print("Output tensor shape:", y.shape)
print("\nInput features:")
print(X)
print("\nTarget values:")
print(y)
```

## Cell 4

```python
X_mean = X.mean(dim=0)
X_std = X.std(dim=0)

y_mean = y.mean()
y_std = y.std()

X_normalized = (X - X_mean) / X_std
y_normalized = (y - y_mean) / y_std

model = nn.Sequential(
    nn.Linear(2, 16),
    nn.ReLU(),
    nn.Linear(16, 1)
)

print("Neural network created successfully!")
print(model)
```

## Cell 5

```python
import torch.optim as optim

loss_function = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

for epoch in range(1000):
    predictions = model(X_normalized)
    loss = loss_function(predictions, y_normalized)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print("Training completed!")
print("Final loss:", round(loss.item(), 6))
```

## Cell 6

```python
model.eval()

with torch.no_grad():
    predictions = model(X_normalized)
    predictions = predictions * y_std + y_mean

print("Actual temperatures:")
print(y.squeeze().tolist())

print("\nPredicted temperatures:")
print(predictions.squeeze().tolist())

mae = torch.mean(torch.abs(predictions - y))

print("\nMean Absolute Error:", round(mae.item(), 4), "°C")
```

## Cell 7

```python
new_weather = torch.tensor([[33.0, 70.0]])

new_weather_normalized = (new_weather - X_mean) / X_std

model.eval()

with torch.no_grad():
    prediction = model(new_weather_normalized)
    prediction = prediction * y_std + y_mean

print("Actual temperature: 33°C")
print("Relative humidity: 70%")
print("Predicted feels-like temperature:",
      round(prediction.item(), 2), "°C")
```
